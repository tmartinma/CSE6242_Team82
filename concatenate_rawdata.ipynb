```python
from pathlib import Path
import pandas as pd
import pyarrow
import fastparquet
```

```python
# Project root
base_dir = Path("data")

# Input folders
green_folder = base_dir / "green_taxi_2025"
yellow_folder = base_dir / "yellow_taxi_2025"
```

```python
# Output folder
output_folder = base_dir / "concate_data"
output_folder.mkdir(parents=True, exist_ok=True)
```

```python
def combine_parquet_files(folder: Path, pattern: str, output_file: Path):
    files = sorted(folder.glob(pattern))

    if not folder.exists():
        print(f"Folder does not exist: {folder.resolve()}")
        return

    if not files:
        print(f"No files found for pattern: {pattern} in {folder.resolve()}")
        return

    print(f"\nFound {len(files)} files in {folder.name}:")
    for f in files:
        print(" -", f.name)

    df = pd.concat(
        (pd.read_parquet(f, engine="pyarrow") for f in files),
        ignore_index=True
    )

    df.to_parquet(output_file, engine="pyarrow", index=False)

    print(f"\nSaved combined file to: {output_file}")
    print(f"Shape: {df.shape}")

```

```python
# Combine green taxi files
combine_parquet_files(
    folder=green_folder,
    pattern="green_tripdata_2025-*.parquet",
    output_file=output_folder / "green_tripdata_2025_all.parquet"
)

```

```python
# Combine yellow taxi files
combine_parquet_files(
    folder=yellow_folder,
    pattern="yellow_tripdata_2025-*.parquet",
    output_file=output_folder / "yellow_tripdata_2025_all.parquet"
)
```

```python
# Define the exact path where the yellow taxi data was saved
file_path = Path("data/concate_data/yellow_tripdata_2025_all.parquet")

# Read the data into a DataFrame
df_yellow = pd.read_parquet(file_path, engine="pyarrow")

# Check the shape and the first few rows to verify it loaded correctly
print(f"Shape: {df_yellow.shape}")
display(df_yellow.head())
```

```python
# ── Combine yellow + green into one unified parquet ──────────────────────────
yellow_path = output_folder / "yellow_tripdata_2025_all.parquet"
green_path  = output_folder / "green_tripdata_2025_all.parquet"

print("Loading yellow...")
df_yellow = pd.read_parquet(yellow_path, engine="pyarrow")
df_yellow["taxi_type"] = "yellow"
print(f"  Yellow shape: {df_yellow.shape}")

print("Loading green...")
df_green = pd.read_parquet(green_path, engine="pyarrow")
df_green["taxi_type"] = "green"
print(f"  Green shape:  {df_green.shape}")

# Keep only shared columns (yellow/green have slightly different column names)
shared_cols = sorted(set(df_yellow.columns) & set(df_green.columns))
print(f"\nShared columns ({len(shared_cols)}): {shared_cols}")

df_combined = pd.concat(
    [df_yellow[shared_cols], df_green[shared_cols]],
    ignore_index=True
)

print(f"\nCombined shape: {df_combined.shape}")

combined_path = output_folder / "combined_tripdata_2025_all.parquet"
df_combined.to_parquet(combined_path, engine="pyarrow", index=False)
print(f"Saved to: {combined_path}")

display(df_combined.head())
```
