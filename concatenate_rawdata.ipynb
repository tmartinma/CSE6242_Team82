```python
from pathlib import Path
import pandas as pd
import pyarrow
import fastparquet
```

```python
# 1) green taxi path
folder = Path("data/green_taxi_2025")
# 2) Get all green trip parquet files in order
files = sorted(folder.glob("green_tripdata_2025-*.parquet"))

if not files:
    raise FileNotFoundError(f"No matching parquet files found in {folder.resolve()}")

print(f"Found {len(files)} source files:")
for f in files:
    print(f.name)
```

```python
combined_df = pd.concat((pd.read_parquet(f) for f in files), ignore_index=True)

output_file = folder / "green_tripdata_2025_all.parquet"
combined_df.to_parquet(output_file, index=False)

print(f"\nSaved to: {output_file}")
print("Shape:", combined_df.shape)
```
